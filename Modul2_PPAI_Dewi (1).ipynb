```python
print("Praktikum Pemrograman Artificial Intelligence - Variable dan Tipe Data")

# tipe data boolean
a = False
print(a)
print(type(a))

# tipe data string
b = "Ini adalah tulisan berupa String"
print(b)

# tipe data integer
c = 100
print(c)

# tipe data float
d = 0.001
print(d)

# bilangan heksadesimal
e = 0x01
print("bilangan desimal dari 0x01 adalah", e)

# tipe data integer
f = 10
print(f)

# tipe data complex
g = 2 + 6j
print(type(g))
print(g)

h = 2 + 6j
print(type(h))

# tipe data list
i = [96, 97, 98, 99, 100]
print(i)

j = ["seratus", "dua ratus", "tiga ratus"]
print(j)

# tipe data dictionary
k = {"nama": "Dewi", "umur": 18}
print(k)

# string dengan tanda petik satu di dalamnya
l = "This string contains a single quote (') character."
print(l)
```
